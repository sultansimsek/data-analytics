# Exercise 1.C Performing operations on arrays

```python
r_1 = np.random.randint(1, 20, size=2)

r_2 = np.random.randint(1, 20, size=(1,2))

r_3 = np.random.rand(2,1)

r_4 = np.random.randint(20, 100, size=(4,5))

r_5 = np.random.rand(4,5)
```

## Permutations

```python
print(r_4)
```

```python
np.random.shuffle(r_4)

print(r_4)
```

Shuffle changes the order of rows inside the original array.
The original array itself is modified.

```python
new_shuffle = np.random.permutation(r_4)

print(new_shuffle)

print(r_4)
```

shuffle changes the original array directly,
while permutation creates a shuffled copy
without changing the original array.


## Mathematical Operations

```python
print(r_1)
print(r_1.shape)

print(r_2)
print(r_2.shape)

result = np.add(r_1, r_2)

print(result)
print(result.shape)
```

```python
print(r_1)
print(r_1.shape)

print(r_3)
print(r_3.shape)

result = np.add(r_1, r_3)

print(result)
print(result.shape)
```

```python
print(r_2)
print(r_2.shape)

print(r_4)
print(r_4.shape)

result = np.add(r_2, r_4)

print(result)
print(result.shape)
```

```python
np.add(r_1, r_4)

# Error explanation:
# r_1 and r_4 cannot broadcast together because
# their shapes are incompatible.
```

```python
print(np.add(np.array([1]), r_4))
```

```python
print(np.add(np.array([[1]]), r_4))
```

```python
print(np.add(np.array([1,1,1,1,1]), r_4))
```

```python
print(np.add(np.array([[1],[1]]), r_4))

# Error explanation:
# The shapes do not align properly for broadcasting.
```

Broadcasting allows NumPy to perform operations
on arrays with different shapes when the dimensions
are compatible.

Smaller arrays are stretched across larger arrays.
The resulting array usually takes the larger shape.

## Grocery Problem

```python
# [vegetables, fruits, snacks, beverages]

bag1 = np.array([2,3,4,0])

bag2 = np.array([4,0,1,2])

print(bag1)
print(bag2)
```

```python
print(
    f"The first bag contains {np.sum(bag1)} items, "
    f"and the second bag contains {np.sum(bag2)} items."
)
```

```python
totals = bag1 + bag2

print(
    f"There are {totals[0]} vegetables, "
    f"{totals[1]} fruits, "
    f"{totals[2]} boxes of snacks, "
    f"and {totals[3]} beverages in the two grocery bags."
)
```
