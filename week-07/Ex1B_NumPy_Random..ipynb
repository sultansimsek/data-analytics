## Generating Random Numbers


```python
import numpy as np
```

```python
r_1 = np.random.randint(1, 20, size=2)
print(r_1)
```

```python
r_2 = np.random.randint(1, 20, size=(1,2))
print(r_2)
```

```python
r_3 = np.random.rand(2,1)
print(r_3)
```

```python
r_4 = np.random.randint(20, 100, size=(4,5))
print(r_4)
```

```python
r_5 = np.random.rand(4,5)
print(r_5)
```

## Generating Random Choices from an Array

```python
students = np.array([
    "Jack",
    "Sue",
    "Maddie",
    "Nick",
    "Theo",
    "Eric",
    "Hawa",
    "Gavin",
    "Zeke",
    "Maddy",
    "Karina",
    "Josh"
])
```

```python
call_on = np.random.choice(students)
print(call_on)
```

```python
breakout_groups = np.random.choice(students, size=(3,4))
print(breakout_groups)

#Some students may appear multiple times, while others may not appear at all; so it is not the best way.
```

```python
call_on_weighted = np.random.choice(
    students,
    size=5,
    p=[0.30, 0.10, 0.10, 0.10, 0.10, 0.05,
       0.05, 0.05, 0.05, 0.05, 0.05, 0.00]
)

print(call_on_weighted)


# Student 1 has a 30% chance.
# Student 12 has a 0% chance.
# Other students were assigned smaller probabilities.
```

## Generating a Normal Distribution

```python
bell_curve = np.random.normal(loc=10, scale=5, size=(4,5))
print(bell_curve)
```
