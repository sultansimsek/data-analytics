```python
import pandas as pd
import numpy as np
```

## Series

```python
# Plain Python objects

aa = [1, 2, 3, 4]

ab = (5, 6, 7, 8)

ac = {9, 10, 11, 12}

ad = {
    'first': 13,
    'second': 14,
    'third': 15,
    'fourth': 16
}
```

```python
# Create Series objects

ser1 = pd.Series(aa)

ser2 = pd.Series(ab)

ser3 = pd.Series(ac)

ser4 = pd.Series(ad)

print(ser1)
print()
print(ser2)
print()
print(ser3)
print()
print(ser4)
```

```python
ser5 = pd.Series(17, 18, 19, 20)

# TypeError:
# Series.__init__() takes from 1 to 6 positional arguments but 5 were given
```

## DataFrames from Series

```python
df1 = pd.DataFrame(ser1)

df1
```

```python
df1.insert(1, "Column_2", ser2)

df1.insert(2, "Column_3", ser3)

df1
```

```python
df1["Column_4"] = ser4

df1

# Explanation:
# ser4 uses dictionary keys as index labels instead of numeric indexes.
# Because the indexes do not match df1's numeric indexes,
# pandas fills unmatched rows with NaN values.
```

```python
df1.loc[4] = [5, 9, 12, 0]

df1
```

```python
df2.describe()
```

describe() provides useful statistics such as mean, minimum,
maximum, and standard deviation for numeric columns.

This helps identify data ranges and possible outliers.

Statistics for non-numeric columns are not included,
so categorical information is less useful here.

```python
df3 = pd.read_csv("northwind_example.csv")

df3
```

```python
df3.describe()
```

The describe() results summarize the numeric columns in the dataset.

Compared to the previous dataset, this dataset may contain
different numeric ranges and distributions.

## Coding a DataFrame

```python
test_data = {
    'ID': [1, 2, 3, 4, 5, 6, 7, 8],

    'Name': ['Alicia', 'Bob', 'Charlie', 'Dominic',
             'Eve', 'Frank', 'Grace', 'Heidi'],

    'Age': [25, np.nan, 22, 19, 31, 35, np.nan, 18],

    'City': ['NY', 'LA', 'Chicago', 'Houston',
             'Phoenix', 'Boston', 'Austin', 'San Diego'],

    'Score': [85, 92, -1, 65, 78, np.nan, 55, 90]
}
```

```python
df4 = pd.DataFrame(test_data)

df4
```

```python
df4.to_csv("test_data.csv", index=False)
```
