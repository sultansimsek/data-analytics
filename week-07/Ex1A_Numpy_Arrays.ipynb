# Exercise 1.A NumPy: Manipulating Arrays

```python
import numpy as np
```

## Looping Through Arrays

```python
arrA = np.array([
    [11, 12, 13],
    [14, 15, 16]
])

print(arrA)
```

```python
for x in arrA:
    print(x * 2)
```

```python
for x in arrA:
    for y in x:
        print(y * 2)
```

The first loop works with entire rows of the array,
so multiplying by 2 affects all numbers in the row at once.

The second loop goes deeper into each row and works
with individual values one at a time.

```python
for x in arrA:
    for y in x:
        loop_result = x * 2

        print(loop_result)
```

## Joining and Stacking Arrays

```python
joined_array_1 = np.concatenate((arrA, arrA), axis=0)

print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis=1)

print(joined_array_2)
```

```python
stacked_array_1 = np.stack((joined_array_1, joined_array_1))
print(stacked_array_1)
```

```python
stacked_array_ = np.stack((joined_array_1, joined_array_2))
print(stacked_array_2)
```

### This gives an error because both arrays must have the same shape to be stacked together.

```python
another_joined_array = np.concatenate((joined_array_1, arrA))
print(another_joined_array)
```

```python
another_2 = np.concatenate((joined_array_1, arrA), axis=1)
print(another_2)
```

### This fails because axis=1 means combining columns, but the arrays do not have matching row sizes.

```python
h_result = np.hstack((arrA, arrA))
print(h_result)

# hstack joins arrays horizontally (side-by-side)

v_result = np.vstack((arrA, arrA))
print(v_result)

# vstack joins arrays vertically (top-to-bottom)

d_result = np.dstack((arrA, arrA))
print(d_result)

# dstack stacks arrays depth-wise (creates a 3D array)
```

## Splitting Arrays

```python
splitsky = np.array_split(joined_array_1, 2)
print(splitsky)

# This creates 2 arrays.
# Each resulting array is still 1-dimensional.
```

```python
spliterino = np.array_split(joined_array_1, 3)
print(spliterino)

# This creates 3 arrays.
# Each resulting array is also 1-dimensional.
```

## Searching and Replacing Array Values

```python
result_1 = np.where(arrA % 2 == 0, arrA, 'odd')
print(result_1)
```
